# Difference b/w List and Numpy_Array

```python
# SPEED DIFF B/W LIST AND NUMPY ARRAY

import numpy as np
import time
a1=np.array([12,13,14,15])
print(a1)
a1

a=[1,2,3,4]*10000
b=np.array(a)
start=time.time()
new_b=b*2
print("time taken for numpy array:",time.time()-start)

a=[1,2,3,4]*10000
strt=time.time()
for i in range(len(a)):
  a[i]=a[i]*2
print("time taken for lists:", time.time()-strt)


```

```python
# MEMORY DIFFERENCE B/W LIST AND NUMPY ARRAY

import sys
a=[1,2,3,4,5]
arr=np.array(a)
print("list size:",sys.getsizeof(a))
print("array size:",arr.nbytes)
```

```python
# NUMPY ALLOWS SAME DATATYPE

a=[1,2.4,"for you"]
arr=np.array(a)
print(a,type(a))
print(arr,type(arr))
```

# BUILT-IN FUNCTIONS

```python
# DIIMENSIONS AND SHAPE OF AN ARRAY

#shpae =(no.of_elemnes in row,)
# shape= (row, column)
# shape=(layer/boxes,rows,columns)

scl=np.array(17)
print("dimension od scalar:",scl.ndim)
print("shape of the scl:",scl.shape)
arr=np.array([12,70])#1D
print("dimension of arr:",arr.ndim)
print("shape of the array:",arr.shape)
arr2=np.array([[16,17,18],[19,20,21]])#2D
print("Dimension of the arr2:",arr2.ndim)
print("shape of the arr2:",arr2.shape)
arr3=np.array([[[13,12],[14,15]]])# 3D
print("Dimension of arr3:",arr3.ndim)
print("shape of the arr3:",arr3.shape)
```

```python
# ASSIGNING VALUES IN ARRAY

arr =np.arange(1,20) # (start,stop)
arr2=np.arange(1,20,2) #(start,stop,step)
print(arr)
print(arr2)
```

```python
# linspace()- it si used when assigning specific no.of values between two numbers in an array.

arr=np.linspace(0,1,5)
print(arr)

arr2= np.linspace(10,20,30) # (start,stop, no.of values between them)
print(arr2)
```

```python
# logspace()- it creates numbers that are evenly spaced on log scale.
# default base - 10

a=np.logspace(2,1,5)
print(a)

b=np.logspace(1,5,3) # (start_exponent, stop_exponent, no.of_values)
print(b)

c= np.logspace(1,5,num=4,base=5)
print(c)
```

```python
#np.zeros -> creates an array full of zeroe values.
#import numpy as np
arr= np.zeros(10 ,dtype=int)
print(arr)

a=np.zeros([2,5])
print(a)

# np.ones-> does same as zeros , but fill the array with ones

one=np.ones(3)
print(one)

one_matrix=np.ones([3,2],dtype=int)
print(one_matrix)



# np.full() -> it is used to fill the array with same value

defa_val=np.full([2,3],'jk') #(no.of_values, value_filled_in_array)
print(defa_val)

ad=np.full(7,7)
print(ad)



```

```python
#np.empty ->unintialized array i.e it creates an array without setting/assigning any values in it.

arr=np.empty([2,3])
print(arr)
```

```python
# np.random.rand()-> it produce random float values (0 to 1)
arr= np.random.rand(2,4)
print(arr)

# np.random.randn -> it generate float values which can be -ve , mostly near to zero , and it based on normal distribution

arr_n=np.random.randn(2,4)
print(arr_n)

#np.random.randint() -> produce random integer values

arr_int=np.random.randint(13,200) #(start,stop)-> they randomly choose a value b/w them
arr3=np.random.randint(2,24,[2,3])
print(arr_int)
print(arr3)
```
